```python


# Define Hubble constant and conformal time range
H = 1.0  # Set H = 1 for simplicity
eta = np.linspace(-10, -0.1, 1000)  # Conformal time must be negative

# Compute scale factors
a_expanding = -1 / (H * eta)  # Expanding de Sitter
a_contracting = 1 / (H * -eta)  # Time-reversed (contracting) de Sitter

a_expanding2 = -1 / (H * eta)  # Expanding de Sitter
a_contracting2 = 1 / (H * eta)  # Time-reversed (contracting) de Sitter


# Plot
plt.figure(figsize=(8, 5))
plt.plot(eta, a_expanding, label=r'Expanding: $a(\eta) = -\frac{1}{H\eta}$', color='blue')
plt.plot(-eta, a_contracting, label=r'Contracting: $a(\eta) = \frac{1}{H\eta}$', color='red', linestyle='dashed')


plt.plot(-eta, a_expanding, label=r'Expanding2: $a(\eta) = -\frac{1}{H\eta}$', color='green')
plt.plot(eta, a_contracting, label=r'Contracting:2 $a(\eta) = \frac{1}{H\eta}$', color='cyan', linestyle='dashed')


# Labels and title
plt.xlabel(r'Conformal time $\eta$')
plt.ylabel(r'Scale factor $a(\eta)$')
plt.title('Scale Factor in de Sitter Space (Expanding & Contracting)')
plt.axhline(0, color='black', linewidth=0.5)
plt.legend()
#plt.grid()

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define constants
H = 1.0  # Hubble constant (arbitrary units, set to 1 for simplicity)

# Time array (from -10 to 10, in arbitrary units)
t = np.linspace(0, 10, 1000)
t_back = np.linspace(-10, 0, 1000)

# Scale factor for de Sitter expansion: a(t) = e^(Ht)
a_t = np.exp(H * t)

# Time-reversed scale factor: a(-t) = e^(-Ht)
a_t_reversed = np.exp(-H * t_back)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(t, a_t, label=r'$a(t) = e^{Ht}$ (Expanding)', color='blue')
plt.plot(t_back, a_t_reversed, label=r'$a(-t) = e^{-Ht}$ (Contracting)', color='red')

# Add labels and title
plt.xlabel('Time (t)')
plt.ylabel('Scale Factor (a)')
plt.title('Scale Factor in de Sitter Phase')
plt.legend()

# Set logarithmic scale for y-axis due to exponential growth/decay
plt.yscale('log')

# Add grid for better readability
#plt.grid(True, which="both", ls="--")

# Show the plot
plt.show()
```
